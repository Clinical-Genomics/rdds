# Visualize 300+ performance

Generate plots for visualizing 300 case performance and effects of thresholding.

Used on the tarball containing validation metrics for genmod, mivmir and gicam.

```python
import pandas as pd
```

```python
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
ls pre-validation-cases
```

```python
df_genmod = pd.read_csv('pre-validation-cases/genmod-performance.csv')
```

```python
df_gicam = pd.read_csv('pre-validation-cases/gicam-performance.csv')
```

## Define bound on recall

```python
recall_constraint = 'recall_scores>=0.927'
```

```python
df_gicam.query(recall_constraint).mcc_scores.max()/ df_genmod.query(recall_constraint).mcc_scores.max()
```

```python
df_gicam.query(recall_constraint).precision_scores.max()/ df_genmod.query(recall_constraint).precision_scores.max()
```

```python
df_gicam.query(recall_constraint).thresholds.max(), df_genmod.query(recall_constraint).thresholds.max()
```

```python
optimal_idx_gicam = df_gicam.mcc_scores.argmax()
optimal_idx_genmod = df_genmod.mcc_scores.argmax()
recall_constraint_gicam = df_gicam.query(recall_constraint).thresholds.argmax()
recall_constraint_genmod = df_genmod.query(recall_constraint).thresholds.argmax()

fig = plt.figure(figsize=(25,15), dpi=400)
ax, ax1, ax2, ax3 = fig.subplots(1, 4)
ax.plot(df_genmod.thresholds, df_genmod.f1_scores)
ax.plot(df_gicam.thresholds, df_gicam.f1_scores)
ax.scatter(df_genmod.thresholds.iloc[optimal_idx_genmod],  df_genmod.f1_scores.iloc[optimal_idx_genmod])
ax.scatter(df_gicam.thresholds.iloc[optimal_idx_gicam],  df_gicam.f1_scores.iloc[optimal_idx_gicam])
ax.scatter(df_genmod.thresholds.iloc[recall_constraint_genmod],  df_genmod.f1_scores.iloc[recall_constraint_genmod], marker='x')
ax.scatter(df_gicam.thresholds.iloc[recall_constraint_gicam],  df_gicam.f1_scores.iloc[recall_constraint_gicam], marker='x')
ax.legend(['F1 Genmod', 'F1 Gicam'])
ax.set_ylim(-0.1, 1.1)
ax.grid()
ax1.plot(df_genmod.thresholds, df_genmod.mcc_scores)
ax1.plot(df_gicam.thresholds, df_gicam.mcc_scores)
ax1.scatter(df_genmod.thresholds.iloc[optimal_idx_genmod],  df_genmod.mcc_scores.iloc[optimal_idx_genmod])
ax1.scatter(df_gicam.thresholds.iloc[optimal_idx_gicam],  df_gicam.mcc_scores.iloc[optimal_idx_gicam])
ax1.scatter(df_genmod.thresholds.iloc[recall_constraint_genmod],  df_genmod.mcc_scores.iloc[recall_constraint_genmod], marker='x')
ax1.scatter(df_gicam.thresholds.iloc[recall_constraint_gicam],  df_gicam.mcc_scores.iloc[recall_constraint_gicam], marker='x')
ax1.legend(['MCC Genmod', 'MCC Gicam'])
ax1.set_ylim(-0.1, 1.1)
ax1.grid()
ax2.plot(df_genmod.thresholds, df_genmod.recall_scores)
ax2.plot(df_gicam.thresholds, df_gicam.recall_scores)
ax2.scatter(df_genmod.thresholds.iloc[optimal_idx_genmod],  df_genmod.recall_scores.iloc[optimal_idx_genmod])
ax2.scatter(df_gicam.thresholds.iloc[optimal_idx_gicam],  df_gicam.recall_scores.iloc[optimal_idx_gicam])
ax2.scatter(df_genmod.thresholds.iloc[recall_constraint_genmod],  df_genmod.recall_scores.iloc[recall_constraint_genmod], marker='x')
ax2.scatter(df_gicam.thresholds.iloc[recall_constraint_gicam],  df_gicam.recall_scores.iloc[recall_constraint_gicam], marker='x')
ax2.legend(['Recall Genmod', 'Recall Gicam'])
ax2.set_ylim(-0.1, 1.1)
ax2.grid()
ax3.plot(df_genmod.thresholds, df_genmod.precision_scores)
ax3.plot(df_gicam.thresholds, df_gicam.precision_scores)
ax3.scatter(df_genmod.thresholds.iloc[optimal_idx_genmod],  df_genmod.precision_scores.iloc[optimal_idx_genmod])
ax3.scatter(df_gicam.thresholds.iloc[optimal_idx_gicam],  df_gicam.precision_scores.iloc[optimal_idx_gicam])
ax3.scatter(df_genmod.thresholds.iloc[recall_constraint_genmod],  df_genmod.precision_scores.iloc[recall_constraint_genmod], marker='x')
ax3.scatter(df_gicam.thresholds.iloc[recall_constraint_gicam],  df_gicam.precision_scores.iloc[recall_constraint_gicam], marker='x')
ax3.legend(['Precision Genmod', 'Precision Gicam'])
ax3.set_ylim(-0.1, 1.1)
ax3.grid()
fig.suptitle(f'\nx: {recall_constraint}, o: max(MCC)\n Recall Constraints Thresholds: Genmod:{df_genmod.thresholds.iloc[recall_constraint_genmod]:.3f}, GICAM:{df_gicam.thresholds.iloc[recall_constraint_gicam]:.3f}')
fig.tight_layout()
```
